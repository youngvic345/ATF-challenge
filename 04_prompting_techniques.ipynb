## Part 1: Prompting Fundamentals

### 1.1 Zero-Shot Prompting

```python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just finished work, had dinner, and now...',
        'expected': 'leisure/relaxation activity'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from: I love outdoor activities but hate crowded places',
        'expected': {'likes': ['outdoor', 'quiet'], 'dislikes': ['crowds']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
```

**Task**: Design prompts for your digital twin use case.

```python
# ==========================================================
# EXERCISE 1.1: ZERO-SHOT PROMPTS FOR DIGITAL TWIN
# ==========================================================
my_prompts = [
    {
        'task': 'activity_prediction',
        'prompt': 'Context: It is 6:00 PM on a Tuesday, workday tasks are completed, and energy is high. Predict the user\'s next activity and location.',
        'expected': 'Heading to Parklands Sports Club for evening workout/fitness training.'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences and constraints from: "I want a quiet place to work on Python and neural networks, with fast Wi-Fi and good black coffee, avoiding noisy cafes."',
        'expected': {
            'likes': ['quiet workspace', 'fast Wi-Fi', 'black coffee', 'coding environments'],
            'dislikes': ['noisy cafes', 'crowded areas']
        }
    },
    {
        'task': 'schedule_decision',
        'prompt': 'Given the choice between an unscheduled evening meeting and a planned sports workout, how should the digital twin prioritize if the calendar is otherwise free?',
        'expected': 'Protect the workout routine unless the meeting is marked urgent or high-priority.'
    },
    {
        'task': 'persona_response',
        'prompt': 'User message: "Can you summarize our plan for this weekend\'s tech build?" Reply in the user\'s direct, concise, and structured voice.',
        'expected': 'Bullet-pointed execution plan focused on model deployment, milestones, and testing.'
    }
]

# Display designed prompts
print(f"Configured {len(my_prompts)} Digital Twin Prompts:\n")
for i, item in enumerate(my_prompts, 1):
    print(f"[{i}] Task: {item['task']}")
    print(f"    Prompt:   {item['prompt']}")
    print(f"    Expected: {item['expected']}\n")
```
